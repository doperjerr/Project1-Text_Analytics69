## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |

| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่


## แบบฟอร์ม Insight Card

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

> **Insight:** ประโยคเดียวที่บอกสิ่งที่พบ ต้องมีตัวเลขอยู่ในประโยค \
> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \
> **So what:** กระทบธุรกิจอย่างไร ถ้าประมาณขนาดผลกระทบเป็นตัวเลขได้ยิ่งดี \
> **Action:** ทีมไหนควรทำอะไร และจะรู้ได้อย่างไรว่าได้ผล \
> **ความมั่นใจ:** สูง กลาง หรือ ต่ำ พร้อมเหตุผลสั้น ๆ

**ตัวอย่างที่ยังไม่ใช่ insight**
> ลูกค้าพูดถึงเรื่องราคาบ่อยที่สุด คิดเป็น 32% ของรีวิว

เป็นเพียงข้อสังเกต ไม่ได้บอกว่าใครต้องทำอะไร และ 32% นั้นมากหรือน้อยก็ไม่รู้เพราะไม่มีอะไรให้เทียบ

**ตัวอย่างที่ใช้ได้**
> **Insight:** รีวิวที่พูดถึงบรรจุภัณฑ์รั่วมีเพียง 6.2% ของทั้งหมด (n = 3,100) แต่ได้ดาวเฉลี่ย 2.1 เทียบกับค่าเฉลี่ยรวม 4.0 และกระจุกอยู่ในสินค้า 12 ชิ้นที่ใช้ขวดแบบปั๊ม \
> **หลักฐาน:** ตารางค่าเฉลี่ยดาวแยกตามด้าน และกราฟแท่งสินค้า 12 ชิ้นในผลวิเคราะห์ \
> **So what:** สินค้า 12 ชิ้นนี้คิดเป็น 18% ของยอดรีวิวทั้งหมวด ถ้าแก้ปัญหาฝาขวดได้ คาดว่าคะแนนเฉลี่ยของหมวดจะขยับขึ้นราว 0.15 ดาว \
> **Action:** ทีม Product ตรวจสอบซัพพลายเออร์ฝาขวดของสินค้า 12 รายการนี้ภายในไตรมาสหน้า แล้วติดตามสัดส่วนรีวิวที่มีคำว่า leak, spill, broken cap ซึ่งตอนนี้อยู่ที่ 14% ต่อเดือน \
> **ความมั่นใจ:** กลาง ยืนยันได้จากข้อความ แต่ยังไม่มีข้อมูลการคืนสินค้ามาช่วยยืนยัน

  2. จากรีวิวเชิงลบของสินค้ากลุ่มนั้น ปัญหาเชิง Technical หรือ Usability 3 อันดับแรกที่ส่งผลกระทบต่อคะแนนมากที่สุดคืออะไร?

  2. **ด้านเสียงรบกวน (Noise Level):** ระบบทำงานหรือมอเตอร์ส่งเสียงดังรบกวนการใช้งาน
  3. **ด้านความยุ่งยากในการใช้งาน (Usability):** ติดตั้งยาก คู่มือไม่ชัดเจน หรือขั้นตอนการใช้วุ่นวาย

### 2.2 เตรียมข้อมูลที่เชื่อถือได้

ทำเพียงสิ่งจำเป็นก่อนเริ่มวิเคราะห์

* สุ่มรีวิวอย่างน้อย **20,000 รายการ** และอย่าใช้เพียง N บรรทัดแรก
* เชื่อม review กับ product metadata ด้วย `parent_asin`
* รายงานสั้น ๆ ว่า join สำเร็จกี่เปอร์เซ็นต์ และมี missing/duplicate สำคัญหรือไม่
* เขียน 2–3 บรรทัดว่า **ข้อมูลนี้เป็นตัวแทนของใคร และไม่เป็นตัวแทนของใคร**

> จุดประสงค์คือให้รู้ว่าข้อมูลที่ใช้เชื่อถือได้พอสำหรับคำถามของทีม ไม่ใช่ทำ data cleaning ทุกอย่างที่เป็นไปได้


````python
# ----------------- Your code here -----------------
# อ่าน Metadata
df_meta = pd.read_json(meta_path, lines=True)

print("จำนวนสินค้า:", len(df_meta))
display(df_meta.head())

# สุ่ม Review มา 20,000 รีวิว

def iter_jsonl(path):
    with open(path, encoding='utf-8') as f:
        for line in f:
            yield json.loads(line)

# นับจำนวนรีวิวทั้งหมด
total_reviews = 0

for row in iter_jsonl(review_path):
    total_reviews += 1

print("จำนวนรีวิวทั้งหมด:", total_reviews)

# สุ่มตำแหน่งของรีวิว
sample_index = set(
    random.sample(range(total_reviews), SAMPLE_N)
)

# อ่านเฉพาะรีวิวที่สุ่มได้
sample_reviews = []

for i, row in enumerate(iter_jsonl(review_path)):
    if i in sample_index:
        sample_reviews.append(row)

df_review = pd.DataFrame(sample_reviews)

print("จำนวนรีวิวที่สุ่ม:", len(df_review))
display(df_review.head())

df_meta['matched'] = 1

df = pd.merge(
    df_review,
    df_meta,
    on='parent_asin',
    how='left', suffixes=('_review', '_meta')
)

print("จำนวนข้อมูลหลัง merge:", len(df))
display(df.head())

# ดูว่า Join สำเร็จกี่ %
join_percent = df['matched'].notnull().sum() / len(df) * 100

print("Join สำเร็จ:", round(join_percent, 2), "%")

# ตรวจ Missing
print("ตรวจmissing")
print(df[
    ['rating', 'text', 'parent_asin', 'matched']
].isnull().sum())

duplicate = df.duplicated(
    subset=['user_id', 'parent_asin', 'rating', 'text', 'timestamp']
).sum()

print("จำนวนรีวิวซ้ำ:", duplicate)

df = df.drop_duplicates(
    subset=['user_id', 'parent_asin', 'rating', 'text', 'timestamp']
)

print("จำนวนข้อมูลหลังลบรีวิวซ้ำ:", len(df))
````

### 2.3 สำรวจข้อมูลเท่าที่จำเป็น

เลือกดูเฉพาะสิ่งที่ช่วยให้เข้าใจ brief ของทีม เช่น

* การแจกแจงดาว
* verified purchase / helpful vote
* ราคา หรือกลุ่มสินค้า
* เวลา
* ความยาวรีวิว

ทำ **2–4 ตารางหรือกราฟ** ก็พอ แล้วเขียนสั้น ๆ ว่า

> “สิ่งใดจาก EDA ทำให้ทีมอยากวิเคราะห์ต่อ และเพราะอะไร”

ไม่ต้องทำ EDA ทุกตัวแปร


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ast
import re
import textwrap # เพิ่ม library สำหรับช่วยตัดคำขึ้นบรรทัดใหม่

# ==========================================
# 1. ทำความสะอาดและสร้างตัวแปร (Data Cleaning)
# ==========================================
def get_subcat(cat_val):
    if isinstance(cat_val, str):
        try:
            cats = ast.literal_eval(cat_val)
        except:
            return 'Unknown'
    else:
        cats = cat_val

    if isinstance(cats, list):
        if len(cats) > 1: return cats[1]
        elif len(cats) == 1: return cats[0]
    return 'Unknown'

def clean_price(price_val):
    if pd.isna(price_val) or str(price_val).strip().lower() == 'none': return np.nan
    match = re.search(r'\d+\.?\d*', str(price_val).replace(',', ''))
    return float(match.group()) if match else np.nan

df['sub_category'] = df['categories'].apply(get_subcat)
df['clean_price'] = df.get('price_meta', df.get('price')).apply(clean_price)
df['review_length'] = df['text'].astype(str).apply(len)
df['datetime'] = pd.to_datetime(df['timestamp'], unit='ms')
df['year'] = df['datetime'].dt.year

# กรอง Verified Purchase และหมวดหมู่ที่มีรีวิว >= 30
df_valid = df[df['verified_purchase'] == True].copy()
valid_cats = df_valid['sub_category'].value_counts()[df_valid['sub_category'].value_counts() >= 30].index
df_valid = df_valid[df_valid['sub_category'].isin(valid_cats)].copy()

# แบ่งกลุ่ม Sentiment
df_valid['sentiment'] = pd.cut(df_valid['rating'], bins=[0, 2, 3, 5],
                               labels=['Negative (1-2)', 'Neutral (3)', 'Positive (4-5)'])
df_valid['is_negative'] = df_valid['rating'] <= 2

sns.set_theme(style="whitegrid")

# ==========================================
# กราฟ 1: Sentiment Distribution
# ==========================================
plt.figure(figsize=(10, 6))
top_cats_volume = df_valid['sub_category'].value_counts().nlargest(10).index
df_top_stacked = df_valid[df_valid['sub_category'].isin(top_cats_volume)]

cross_tab = pd.crosstab(df_top_stacked['sub_category'], df_top_stacked['sentiment'], normalize='index') * 100
cross_tab = cross_tab.sort_values('Negative (1-2)', ascending=True)

cross_tab.plot(kind='barh', stacked=True, figsize=(10, 6), color=['#d73027', '#fee08b', '#1a9850'])
plt.title('1. Sentiment Distribution by Top Sub-Categories', fontweight='bold')
plt.xlabel('Percentage (%)')
plt.ylabel('Sub-Category')
plt.legend(title='Sentiment', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# ==========================================
# กราฟ 2: Trend of Negative Reviews Over Time
# ==========================================
plt.figure(figsize=(10, 6))
top_5_bad_cats = cross_tab.tail(5).index
df_bad_trend = df_valid[(df_valid['is_negative'] == True) & (df_valid['sub_category'].isin(top_5_bad_cats))]

trend_data = df_bad_trend[df_bad_trend['year'] >= 2015].groupby(['year', 'sub_category']).size().reset_index(name='count')
sns.lineplot(data=trend_data, x='year', y='count', hue='sub_category', marker='o', linewidth=2)
plt.title('2. Trend of Negative Reviews Over Time (Since 2015)', fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Number of 1-2 Star Reviews')
plt.legend(title='Sub-Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# Calculate category statistics for Graph 3
cat_stats = df_valid.groupby('sub_category').agg(
    total_reviews=('rating', 'count'),
    negative_reviews=('is_negative', lambda x: x.sum() if len(x) > 0 else 0),
    avg_price=('clean_price', 'mean')
).reset_index()

cat_stats['negative_rate'] = (cat_stats['negative_reviews'] / cat_stats['total_reviews']) * 100
# Filter out categories with no reviews or no negative reviews to avoid division by zero or NaN issues
cat_stats = cat_stats[cat_stats['total_reviews'] > 0].copy()
cat_stats = cat_stats.dropna(subset=['avg_price']).copy() # Ensure price is not NaN

# ==========================================
# เตรียมข้อมูลสำหรับ กราฟ 3 (เพิ่มการจัดการ Missing Data)
# ==========================================
cat_stats = df_valid.groupby('sub_category').agg(
    total_reviews=('rating', 'count'),
    negative_reviews=('is_negative', lambda x: x.sum() if len(x) > 0 else 0),
    avg_price=('clean_price', 'mean')
).reset_index()

cat_stats['negative_rate'] = (cat_stats['negative_reviews'] / cat_stats['total_reviews']) * 100
cat_stats = cat_stats[cat_stats['total_reviews'] > 0].copy()

# ---- พิมพ์รายงานผลการจัดการ Missing Data ----
total_cats_before = len(cat_stats)
missing_price_cats = cat_stats['avg_price'].isna().sum()
print("\n" + "="*50)
print(f"รายงานการเตรียมข้อมูลสำหรับ Risk Matrix:")
print(f"- จำนวนกลุ่มสินค้าทั้งหมด (Sub-categories): {total_cats_before} กลุ่ม")
print(f"- พบกลุ่มสินค้าที่ไม่มีข้อมูลราคา: {missing_price_cats} กลุ่ม (คิดเป็น {(missing_price_cats/total_cats_before)*100:.1f}%)")
print("=> ทำการ Drop กลุ่มที่ไม่มีข้อมูลราคาออก เพื่อป้องกันความคลาดเคลื่อนในกราฟ")

# Drop missing value ออก
cat_stats = cat_stats.dropna(subset=['avg_price']).copy()
print(f"- ข้อมูลที่พร้อมใช้สำหรับ Risk Matrix: {len(cat_stats)} กลุ่มสินค้า")
print("="*50 + "\n")

# ==========================================
# กราฟ 3: Category Risk Matrix
# ==========================================
plt.figure(figsize=(12, 8))

scatter = sns.scatterplot(data=cat_stats, x='negative_rate', y='avg_price',
                          size='negative_reviews', sizes=(50, 1500),
                          hue='sub_category', alpha=0.7, edgecolor='black')

med_rate = cat_stats['negative_rate'].median()
med_price = cat_stats['avg_price'].median()
plt.axvline(med_rate, color='gray', linestyle='--', alpha=0.8)
plt.axhline(med_price, color='gray', linestyle='--', alpha=0.8)

# ขยายพื้นที่แกน X ด้านขวาเพิ่มอีก 5% เพื่อให้มีที่ว่างสำหรับวางตัวหนังสือ
plt.xlim(cat_stats['negative_rate'].min() - 1, cat_stats['negative_rate'].max() + 5)

# ไฮไลต์พื้นที่วิกฤต (ขยายขอบเขตกล่องแดงให้พอดีกับแกน X ที่ขยายใหม่)
plt.axvspan(med_rate, cat_stats['negative_rate'].max() + 5,
            ymin=(med_price - cat_stats['avg_price'].min()) / (cat_stats['avg_price'].max() - cat_stats['avg_price'].min()),
            ymax=1, color='red', alpha=0.05)

for i in range(len(cat_stats)):
    rate = cat_stats['negative_rate'].iloc[i]
    price = cat_stats['avg_price'].iloc[i]
    cat_name = cat_stats['sub_category'].iloc[i]

    if rate > med_rate and price > med_price:
        # จัดการชื่อที่ยาวเกิน 25 ตัวอักษรให้ปัดขึ้นบรรทัดใหม่
        wrapped_name = '\n'.join(textwrap.wrap(cat_name, width=25))
        detail_label = f"{wrapped_name}\n(Rate: {rate:.1f}%, Price: ${price:.0f})"

        plt.text(rate + 0.3, price, detail_label, color='darkred', weight='bold', fontsize=9, va='center')

plt.title('3. Detailed Risk Matrix: Negative Rate vs Average Price', fontweight='bold')
plt.xlabel('Negative Review Rate (%)')
plt.ylabel('Average Price (USD)')

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
plt.tight_layout()
plt.show()

# ==========================================
# กราฟ 4: Top 10 Critical Products
# แสดง "รหัสสินค้า + Sub-category"
# ==========================================
plt.figure(figsize=(12, 7))

# หากลุ่มสินค้าที่ตกอยู่ในโซนวิกฤต
critical_cats = cat_stats[
    (cat_stats['negative_rate'] > med_rate) &
    (cat_stats['avg_price'] > med_price)
]['sub_category']

df_critical_neg = df_valid[
    (df_valid['is_negative'] == True) &
    (df_valid['sub_category'].isin(critical_cats))
]

# นับจำนวนรีวิว 1-2 ดาวของแต่ละสินค้า
top_products = (
    df_critical_neg
    .groupby(['parent_asin', 'sub_category'])
    .size()
    .reset_index(name='neg_count')
)

# เลือก Top 10 สินค้าที่มีรีวิวเชิงลบมากที่สุด
top_products = (
    top_products
    .sort_values('neg_count', ascending=False)
    .head(10)
)

# สร้างชื่อแกน Y = รหัสสินค้า + Sub-category
top_products['display_label'] = top_products.apply(
    lambda row: f"{row['parent_asin']}\n{row['sub_category']}",
    axis=1
)

# วาดกราฟ
sns.barplot(
    data=top_products,
    x='neg_count',
    y='display_label',
    color='#d73027'
)

plt.title(
    '4. Top 10 Critical Products to Fix First',
    fontweight='bold'
)

plt.xlabel('Number of 1-2 Star Reviews')
plt.ylabel('Product ID & Sub-category')

# ใส่จำนวนรีวิวท้ายแท่ง
for index, value in enumerate(top_products['neg_count']):
    plt.text(
        value + 0.1,
        index,
        str(value),
        va='center',
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
````

### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

### ตัวอย่าง Embedding แบบง่าย

ถ้าสอน `sentence-transformers` มาแล้ว อาจใส่ตัวอย่างสั้น ๆ:

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

texts = [
    "The delivery was very slow.",
    "Shipping took too long.",
    "Excellent sound quality."
]

embeddings = model.encode(texts)

print(embeddings.shape)
```

**รายงานเพียง 3 อย่าง**
1. ทำ preprocessing อะไร และทำไม
2. สร้าง feature อะไรอย่างน้อย 2 ตัว
3. feature เหล่านั้นช่วยตอบคำถามทางธุรกิจอย่างไร

> ไม่ต้องรายงาน vocabulary ก่อน–หลัง เว้นแต่ประเด็นนั้นเกี่ยวข้องกับผลวิเคราะห์

````python
import pandas as pd
import re
from textblob import TextBlob
from sklearn.feature_extraction.text import CountVectorizer

# -----------------------------------------------------------------------------
# 1. Text Preprocessing (ทำความสะอาดข้อความ)
# -----------------------------------------------------------------------------
def clean_text_simple(text):
    text = str(text).lower() # พิมพ์เล็ก
    text = re.sub(r'<.*?>', ' ', text) # ลบ HTML
    text = re.sub(r'http\S+|www\.\S+', '', text) # ลบ Link
    text = re.sub(r'[^a-z0-9\s]', '', text) # ลบเครื่องหมายวรรคตอน (เก็บแค่ a-z และ 0-9)
    return text

df['cleaned_text'] = df['text'].apply(clean_text_simple)
df['cleaned_text'] = df['cleaned_text'].fillna('')

df_negative = df[df['rating'] <= 2].copy()
print(f"จำนวนรีวิวเชิงลบ (1-2 ดาว) ที่จะนำมาวิเคราะห์: {len(df_negative)}")

# 2. ค้นหาปัญหาจากข้อมูลจริงด้วย N-gram Analysis
vectorizer = CountVectorizer(ngram_range=(2, 3), stop_words='english', max_features=15)
X = vectorizer.fit_transform(df_negative['cleaned_text'])

phrases = vectorizer.get_feature_names_out()
frequencies = X.sum(axis=0).A1

df_ngrams = pd.DataFrame({
    'ปัญหาที่ถูกลูกค้าบ่นจริง (Actual Phrases)': phrases,
    'จำนวนครั้งที่พบ': frequencies
}).sort_values(by='จำนวนครั้งที่พบ', ascending=False).reset_index(drop=True)

print("\n--- ผลลัพธ์จากการทำ N-gram (ใช้เป็นฐานในการตั้งสมมติฐาน) ---")
display(df_ngrams)

durability_kw = ['stop', 'stopped', 'broken', 'die', 'died', 'burn', 'last', 'work' ,'stopped working', 'didnt work', 'doesnt work', 'quit working','broke','fail','failed','poor quality','break']
fit_kw = ['fit', 'size', 'wrong', 'small', 'large', 'match','didnt fit', 'doesnt fit', 'does fit']
service_kw = ['service', 'support', 'contact', 'customer service']
price_value_kw = ['refund', 'money', 'waste', 'waste money', 'expensive', 'price', 'value', 'worth']
noise_kw = ['noise','noisy','loud','rattle','buzz','vibration','noisiest']

def check_aspect(text, keywords):
    words = set(text.split())
    if any(kw in words for kw in keywords):
        return 1
    return 0

df_negative['issue_durability'] = df_negative['cleaned_text'].apply(lambda x: check_aspect(x, durability_kw))
df_negative['issue_fit'] = df_negative['cleaned_text'].apply(lambda x: check_aspect(x, fit_kw))
df_negative['issue_service'] = df_negative['cleaned_text'].apply(lambda x: check_aspect(x, service_kw))
df_negative['issue_price'] = df_negative['cleaned_text'].apply(lambda x: check_aspect(x, price_value_kw))
df_negative['issue_noise'] = df_negative['cleaned_text'].apply(lambda x: check_aspect(x, noise_kw))

# -----------------------------------------------------------------------------
# 3. สร้าง Text Features: Sentiment Score (คะแนนความรู้สึก -1.0 ถึง 1.0)
# -----------------------------------------------------------------------------
df_negative['sentiment_score'] = df_negative['cleaned_text'].apply(lambda x: TextBlob(x).sentiment.polarity)

# -----------------------------------------------------------------------------
# 5. แสดงตัวผลลัพธ์ของ Text Features ที่ได้
# -----------------------------------------------------------------------------
feature_columns = [
    'text', 'cleaned_text', 'rating',
    'issue_durability', 'issue_fit', 'issue_service', 'issue_noise','issue_price',
    'sentiment_score'
]

print("--- ตัวอย่าง Text Features ที่ได้จากการปรับปรุงโค้ด ---")
display(df_negative[feature_columns].head())
````

####รายงาน
**preprocessing**

ทำการเปลี่ยนข้อความเป็นตัวพิมพ์เล็กทั้งหมด (Lowercasing) ลบแท็ก HTML ลิงก์ URL และลบเครื่องหมายวรรคตอน (Punctuation) เพื่อลด Noise ในข้อมูล และป้องกันไม่ให้โปรแกรมมองคำเดียวกันเป็นคนละคำเพียงเพราะมีเครื่องหมายติดอยู่ ซึ่งจำเป็นมากต่อการทำ N-gram และ Keyword Matching ให้มีความแม่นยำ

**Text Feature**

1. Aspect Indicators (Binary 0/1): สร้างขึ้นหลังจากทำ N-gram Analysis (CountVectorizer) เพื่อค้นหาปัญหาจริงก่อน แล้วจึงนำชุดคำศัพท์ (Keyword) ที่พบมากที่สุดมาจัดกลุ่มเป็นปัญหา  
2. Sentiment Score (Continuous): ใช้ TextBlob คำนวณขั้วอารมณ์ความรู้สึกของข้อความ โดยมีค่าตั้งแต่ -1.0 (เชิงลบมาก) ถึง 1.0 (เชิงบวกมาก)

**feature เหล่านั้นช่วยตอบคำถามทางธุรกิจอย่างไร**

* Aspect Indicators (คู่กับ N-gram): ช่วยตอบคำถามว่า 'ปัญหาที่ลูกค้าพบจริงๆ คืออะไร' แทนที่จะเดาไปเอง ทีม Product สามารถดูผลรวมของ Indicator และนำงบประมาณไปแก้ปัญหาที่มีจำนวนความถี่สูงสุด (เช่น ปัญหาความทนทาน) ได้อย่างตรงจุดและคุ้มค่าที่สุด")
* Sentiment Score: ช่วยประเมิน 'ความรุนแรงของปัญหา (Severity)' สินค้าที่มีคนให้ 1 ดาวเท่ากัน แต่หากมีค่าเฉลี่ย Sentiment ติดลบมากกว่า (เช่น ลูกค้าโกรธจัดเพราะสินค้าเป็นอันตราย หรือติดต่อขอเคลมไม่ได้) ทีมธุรกิจจะต้องให้ความสำคัญในการระงับการขายหรือเรียกคืน (Recall) สินค้าตัวนั้นก่อนเพื่อรักษาความเชื่อมั่นของแบรนด์

### 2.5 หา Insight ที่ตอบ Client Brief

เลือกเทคนิคจากเมนูด้านบนตามความเหมาะสม **ไม่ต้องทำครบ**

> **Segmentation เป็นทางเลือก ไม่ใช่หัวข้อบังคับ** ทำเมื่อช่วยให้ insight ชัดขึ้นเท่านั้น


#### TECHNIQUE 1: Product-Level Analysis
คำถาม: สินค้ากลุ่มไหน/รหัสอะไรที่มีวิกฤตคะแนน 1-2 ดาวสูงสุด?

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 1. เตรียมข้อมูลรีวิว 1-2 ดาว และแปลง categories
df_low = df[df['rating'] <= 2].copy()
df['categories'] = df['categories'].apply(
    lambda x: ', '.join(x) if isinstance(x, list) else str(x)
)

# 2. คำนวณสัดส่วนรีวิว 1-2 ดาวรายสินค้า
product_summary = (
    df.groupby(['categories', 'main_category', 'parent_asin'])
      .agg(
          Total_Reviews=('rating', 'count'),
          Low_Rating_Reviews=('rating', lambda x: (x <= 2).sum()),
          Average_Rating=('rating', 'mean')
      )
      .reset_index()
)

product_summary['Low_Rating_Rate'] = (
    product_summary['Low_Rating_Reviews'] / product_summary['Total_Reviews'] * 100
)

# 3. กรองเฉพาะสินค้าที่มีรีวิวรวมอย่างน้อย 30 รีวิวเพื่อป้องกัน Bias
product_summary_filtered = product_summary[product_summary['Total_Reviews'] >= 30].copy()

top_problem_products = (
    product_summary_filtered
    .sort_values(['Low_Rating_Rate', 'Low_Rating_Reviews'], ascending=[False, False])
    .head(10)
    .reset_index(drop=True)
)

print("\n===== ผลลัพธ์ Technique 1: Top 10 Problem Products =====")
display(top_problem_products[['parent_asin', 'Total_Reviews', 'Low_Rating_Reviews', 'Low_Rating_Rate']])

# ------------------------------------------------------------
# กราฟสำหรับ Technique 1
# ------------------------------------------------------------
plot_data = top_problem_products.sort_values('Low_Rating_Rate', ascending=True).copy()
plot_data['product_label'] = plot_data['parent_asin'].astype(str)

plt.figure(figsize=(10, 6))

# ไฮไลต์สีแดงเฉพาะ 3 อันดับแรก (ตาม Brief ที่แก้ได้ 3 อย่าง) สีเทาสำหรับอันดับอื่น
colors = ['#C0392B' if i >= len(plot_data) - 3 else '#B8B8B8' for i in range(len(plot_data))]

bars = plt.barh(
    plot_data['product_label'],
    plot_data['Low_Rating_Rate'],
    color=colors,
    height=0.65
)

plt.title('Top 10 Products with Highest Low-Rating Rate\n(Focus on Top 3 for Immediate Action)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Low Rating Rate (%)', fontsize=11)
plt.ylabel('Product ID (parent_asin)', fontsize=11)

# แสดงตัวเลขบนแท่งกราฟ
for bar, rate, count in zip(bars, plot_data['Low_Rating_Rate'], plot_data['Low_Rating_Reviews']):
    plt.text(rate + 0.5, bar.get_y() + bar.get_height()/2,
             f'{rate:.1f}% ({count} reviews)',
             va='center', fontsize=10, fontweight='bold')

plt.xlim(0, plot_data['Low_Rating_Rate'].max() + 15)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
````

#### TECHNIQUE 2: Helpful Vote & Aspect Analysis
คำถาม: ในกลุ่มรีวิว 1-2 ดาวที่คนโหวตว่ามีประโยชน์ ปัญหาหลักคืออะไร?

````python
# 1. กรอง Helpful Vote (ใช้ Percentile 75 และต้องมีคนโหวตอย่างน้อย 1 คน)
q75 = df_low['helpful_vote'].quantile(0.75)
helpful_threshold = max(q75, 1)

negative_helpful = df_low[df_low['helpful_vote'] >= helpful_threshold].copy()
print(f"\nจำนวนรีวิว 1-2 ดาวที่มี Helpful Vote สูง (>= {helpful_threshold} votes): {len(negative_helpful)}")

# 2. ปรับ Keywords ใหม่ให้แยก Service กับ Price ออกจากกัน
issue_keywords_updated = {
    'Durability': ['broke', 'broken', 'break', 'stopped working', 'stop working', 'quit working', 'dead', 'failed', 'failure', 'lasted', 'defective', 'malfunction'],
    'Noise Level': ['noise', 'noisy', 'loud', 'louder', 'buzz', 'rattle', 'hum', 'vibration'],
    'Usability': ['difficult', 'hard to use', 'complicated', 'confusing', 'instructions', 'manual', 'install', 'setup'],
    'Fit & Size': ['fit', 'size', 'small', 'large', 'too small', 'too large', 'tight', 'loose', 'dimension'],
    'Customer Service': ['customer service', 'support', 'contact', 'return', 'replacement'],
    'Price & Value': ['refund', 'money', 'waste', 'expensive', 'price', 'value', 'worth']
}

# 3. ฟังก์ชันตรวจจับปัญหา
def detect_issue(text, keywords):
    text = str(text).lower()
    for keyword in keywords:
        if keyword in text:
            return 1
    return 0

# 4. นับจำนวนปัญหา
for issue, keywords in issue_keywords_updated.items():
    col_name = issue.lower().replace(' & ', '_').replace(' ', '_')
    negative_helpful[col_name] = negative_helpful['cleaned_text'].apply(lambda x: detect_issue(x, keywords))

helpful_issue_result = []
for issue in issue_keywords_updated.keys():
    col_name = issue.lower().replace(' & ', '_').replace(' ', '_')
    count = negative_helpful[col_name].sum()
    percentage = (count / len(negative_helpful) * 100) if len(negative_helpful) > 0 else 0

    helpful_issue_result.append({
        'Issue': issue,
        'Count': int(count),
        'Percentage': percentage
    })

helpful_issue_summary = pd.DataFrame(helpful_issue_result).sort_values('Count', ascending=False).reset_index(drop=True)

print("\n===== ผลลัพธ์ Technique 2: Top Issues in Helpful Negative Reviews =====")
display(helpful_issue_summary)

# ------------------------------------------------------------
# กราฟสำหรับ Technique 2
# ------------------------------------------------------------
# กรองเอาเฉพาะ Top 5 ปัญหา
plot_issue = helpful_issue_summary.head(5).sort_values('Percentage', ascending=True)

plt.figure(figsize=(10, 6))

# แบ่งสีให้เห็นชัดเจน ปัญหาทางกายภาพ (Product) ใช้โทนสีแดง/ส้ม, ปัญหาด้านราคา/บริการใช้โทนเทา
colors = []
for issue in plot_issue['Issue']:
    if issue in ['Durability', 'Fit & Size', 'Noise Level', 'Usability']:
        colors.append('#E67E22') # สีส้มสำหรับสิ่งที่ Product Team ต้องแก้
    else:
        colors.append('#95A5A6') # สีเทาสำหรับ Service / Price

bars = plt.barh(
    plot_issue['Issue'],
    plot_issue['Percentage'],
    color=colors,
    height=0.6,
    edgecolor='white',
    linewidth=1.5
)

plt.title('Top Issues in High-Helpful Negative Reviews\n(Orange = Product Team Responsibility)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel(f'Percentage of Reviews (%) [Total High-Helpful n = {len(negative_helpful)}]', fontsize=11)
plt.ylabel('Reported Issue', fontsize=11)

for bar, value, count in zip(bars, plot_issue['Percentage'], plot_issue['Count']):
    plt.text(value + 0.5, bar.get_y() + bar.get_height()/2,
             f'{value:.1f}% (n={count})',
             va='center', fontsize=10, fontweight='bold')

plt.xlim(0, plot_issue['Percentage'].max() + 15)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
````

````python
# ==========================================
# 2.5 วิเคราะห์เชิงลึก "สินค้าเป้าหมาย × ปัญหา" (Deep Dive) พร้อมสรุป Action Plan เป็นตาราง
# ==========================================
import pandas as pd

# 1. ดึงรหัสสินค้า 3 อันดับแรกที่เราแนะนำให้แก้ (จากเทคนิคที่ 1)
top_3_products = top_problem_products['parent_asin'].head(3)

# 2. เลือกรีวิว 1-2 ดาวของสินค้า 3 ตัวนี้
negative_top3 = df_negative[df_negative['parent_asin'].isin(top_3_products)].copy()

# 3. สรุปจำนวนปัญหาแต่ละประเภท (อิงตามคอลัมน์ที่สร้างจริงในข้อ 2.4)
product_issue_summary = (
    negative_top3
    .groupby(['parent_asin'])
    .agg(
        Total_Negative=('rating', 'count'),
        Durability=('issue_durability', 'sum'),
        Fit_Size=('issue_fit', 'sum'),
        Noise=('issue_noise', 'sum'),
        Service=('issue_service', 'sum'),
        Price=('issue_price', 'sum')
    )
    .reset_index()
)

print("\n===== ตารางเจาะลึกปัญหาของสินค้า Top 3 (จำแนกตามประเภทปัญหา) =====")
display(product_issue_summary)

# ==========================================
# 4. สร้างข้อสรุป Action Plan เป็น "ตาราง (DataFrame)"
# ==========================================
issue_cols = ['Durability', 'Fit_Size', 'Noise', 'Service', 'Price']
action_plan_data = []

# วนลูปอ่านข้อมูลเพื่อหาปัญหาที่เยอะที่สุดของแต่ละสินค้า
for index, row in product_issue_summary.iterrows():
    product_id = row['parent_asin']
    total_neg = row['Total_Negative']

    # หาปัญหาที่คะแนนสูงที่สุด (Max) ในบรรทัดนั้น
    most_common_issue = row[issue_cols].astype(float).idxmax()
    max_count = int(row[most_common_issue])

    # กำหนด Action ให้สอดคล้องกับหน้าที่ของทีม Product
    if most_common_issue in ['Durability', 'Fit_Size', 'Noise']:
        action_text = "🛠️ ทีม Product รับจบ (แก้ปัญหาที่ตัวสินค้า)"
    else:
        action_text = "🤝 ส่งต่อทีมอื่น (ประสาน CS/Pricing)"

    # เก็บข้อมูลเป็นบรรทัดเตรียมสร้างตาราง
    action_plan_data.append({
        'Product_ID': product_id,
        'Total_Negative_Reviews': total_neg,
        'Top_Problem': most_common_issue,
        'Problem_Count': max_count,
        'Action': action_text
    })

# แปลงข้อมูลเป็น Pandas DataFrame
df_action_plan = pd.DataFrame(action_plan_data)

print("\n🎯 ตารางสรุป Action Plan สำหรับทีม Product (เจาะจงรายสินค้า):")
display(df_action_plan)
````

#### TECHNIQUE 3: Severity Analysis (อารมณ์ความรู้สึกตามประเภทปัญหา)

````python
# เพิ่มบรรทัดนี้เพื่อดึง Feature ที่ทำไว้มาใช้งาน
all_negative['sentiment_score'] = df_negative['sentiment_score']

severity_result = []

for issue in issue_keywords_updated.keys():
    col_name = issue.lower().replace(' & ', '_').replace(' ', '_')

    # กรองเฉพาะรีวิวที่มีปัญหานั้นๆ
    df_issue = all_negative[all_negative[col_name] == 1]

    if len(df_issue) > 0:
        mean_sentiment = df_issue['sentiment_score'].mean()
        count_issue = len(df_issue)

        severity_result.append({
            'Issue': issue,
            'Review Count': count_issue,
            'Average Sentiment Score': mean_sentiment
        })

df_severity = pd.DataFrame(severity_result).sort_values('Average Sentiment Score', ascending=True).reset_index(drop=True)

print("\n===== ผลลัพธ์ Technique 3: ความรุนแรงของปัญหา (เรียงจากอารมณ์ติดลบมากที่สุด) =====")
display(df_severity.style.format({
    'Average Sentiment Score': '{:.3f}'
}))

# สร้าง Scatter Plot เพื่อเทียบ ความถี่ (แกน X) vs ความโกรธ (แกน Y)
plt.figure(figsize=(10, 6))

plt.scatter(df_severity['Review Count'], df_severity['Average Sentiment Score'],
            color=['#E67E22' if i in ['Durability', 'Fit & Size', 'Noise Level', 'Usability'] else '#95A5A6' for i in df_severity['Issue']],
            s=200, alpha=0.8)

for i, row in df_severity.iterrows():
    plt.text(row['Review Count'] + 10, row['Average Sentiment Score'], row['Issue'],
             fontsize=10, fontweight='bold')

plt.title('Severity vs Frequency of Issues in Negative Reviews\n(Lower Sentiment Score = More Negative Emotion)', fontweight='bold')
plt.xlabel('Number of Reviews (Frequency)')
plt.ylabel('Average Sentiment Score (Severity)')
plt.gca().invert_yaxis()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

ในตัวอย่างนี้

````python
# ===== Step 1: ตั้งค่า =====
# หากยังไม่มี package:
# !pip install -q requests beautifulsoup4

import re
import time
import requests
import pandas as pd

from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse, urldefrag
from urllib.robotparser import RobotFileParser

# เว็บไซต์ที่เราจะใช้
BASE_URL = "https://www.amarintv.com"
LISTING_URL = "https://www.amarintv.com/news"

# Amarin TV มี URL ข่าวประมาณ
# /news/crime/552772
# /news/social/558145
ARTICLE_PATH_PATTERN = r"^/news/[^/]+/\d+$"

# เริ่มทดลองน้อย ๆ ก่อน
MAX_ARTICLES = 20
DELAY = 1.5
BOT_USER_AGENT = "KKU-TextAnalytics-StudentBot/1.0"

session = requests.Session()
session.headers.update({"User-Agent": BOT_USER_AGENT})

print("Listing page:", LISTING_URL)
print("เป้าหมาย: 1 row = 1 news article")
````

#### Step 2 — ทำความสะอาด URL และตรวจ `robots.txt`

URL เดียวกันอาจเขียนได้หลายแบบ เช่น

`https://www.thaipbs.or.th/news`  
`https://www.thaipbs.or.th/news#main-content`

ส่วนหลัง `#...` เรียกว่า **fragment** และโดยทั่วไปไม่ได้หมายถึงหน้าใหม่ จึงควรตัดออกก่อนตรวจ URL ซ้ำ

จากนั้นตรวจ `robots.txt` ก่อนเข้าแต่ละส่วนของเว็บ ถ้าไม่อนุญาตให้ bot เข้า path นั้น ให้ **เปลี่ยนแหล่งข้อมูล** ไม่ใช่หาวิธีหลบ


````python
# ===== Step 2: normalize URL + robots.txt =====

_robots_cache = {}

def normalize_url(url):
    """ตัด #fragment ออกจาก URL เพื่อไม่ให้หน้าเดียวกันถูกนับซ้ำ"""
    clean_url, _ = urldefrag(url)
    return clean_url.rstrip("/")


def robots_allowed(url, user_agent=BOT_USER_AGENT):
    """คืน True เมื่อ robots.txt อนุญาตให้ bot เข้าถึง URL"""
    p = urlparse(url)
    root = f"{p.scheme}://{p.netloc}"

    if root not in _robots_cache:
        rp = RobotFileParser()
        rp.set_url(urljoin(root, "/robots.txt"))
        try:
            rp.read()
            _robots_cache[root] = rp
        except Exception as e:
            print("อ่าน robots.txt ไม่ได้:", e)
            return False

    return _robots_cache[root].can_fetch(user_agent, url)


print(
    "robots.txt อนุญาตหน้ารวมข่าวหรือไม่:",
    robots_allowed(LISTING_URL)
)
````

#### Step 3 — เข้า “หน้ารวมข่าว” เพื่อหาเฉพาะ Article URLs

ขั้นนี้ **ยังไม่เก็บเนื้อหาข่าว**

Robot เปิดหน้ารวมข่าว แล้วดู `<a href="...">` ทุก link จากนั้นกรองให้เหลือเฉพาะ URL ที่ตรงกับรูปแบบบทความ เช่น

`/news/content/509281`

การกรองนี้คือสิ่งที่ทำให้ Robot ต่างจาก generic crawler ที่วิ่งไปทุกหน้าใน domain


````python
# ===== Step 3: หา link ของ "บทความจริง" =====

def find_article_links(listing_url):
    if not robots_allowed(listing_url):
        raise PermissionError("robots.txt ไม่อนุญาตให้เข้าหน้ารวมข่าวนี้")

    r = session.get(listing_url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")
    article_urls = set()

    for a in soup.find_all("a", href=True):
        url = urljoin(listing_url, a["href"])
        url = normalize_url(url)

        p = urlparse(url)

        # 1) เอาเฉพาะ domain เดียวกับเว็บไซต์หลัก
        same_domain = (p.netloc == urlparse(BASE_URL).netloc)

        # 2) เอาเฉพาะ path ที่มีรูปแบบเป็นบทความข่าว
        is_article = bool(re.match(ARTICLE_PATH_PATTERN, p.path))

        if same_domain and is_article:
            article_urls.add(url)

    return sorted(article_urls)


article_urls = find_article_links(LISTING_URL)

print("พบ article URLs =", len(article_urls))
print("ตัวอย่าง 5 URLs:")

for u in article_urls[:5]:
    print(" -", u)
````

#### Step 4 — เปิดบทความทีละหน้าแล้วดึง “เนื้อหาที่ต้องการ”

ตอนนี้เรามี URL ของบทความแล้ว จึงเปิดทีละข่าวและแยก field ที่จะใช้วิเคราะห์ เช่น

หลักสำคัญคือ **พยายามดึงเฉพาะตัวบทความ** ไม่ใช่ menu/footer/ข่าวแนะนำทั้งหน้า

ตัวอย่างใช้ `<article>` ก่อน ถ้าเว็บไซต์ไม่มี tag นี้จึง fallback ไปที่ `<main>` และนำข้อความจาก `<p>` มาต่อกัน  
เว็บไซต์อื่นอาจต้องเปลี่ยน selector เช่น `.article-body`, `.post-content` หรือ selector เฉพาะของเว็บนั้น


````python
# ===== Step 4: ดึงข้อมูลจากบทความ 1 หน้า =====

def extract_article(url):
    if not robots_allowed(url):
        print("robots.txt ไม่อนุญาต ข้าม:", url)
        return None

    r = session.get(url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")

    # ---- title ----
    h1 = soup.find("h1")
    title = h1.get_text(" ", strip=True) if h1 else ""

    # ---- published date: พยายามอ่านจาก meta หรือ time ----
    published_at = ""

    meta_time = soup.find("meta", attrs={"property": "article:published_time"})
    if meta_time and meta_time.get("content"):
        published_at = meta_time["content"]
    else:
        time_tag = soup.find("time")
        if time_tag:
            published_at = time_tag.get("datetime") or time_tag.get_text(" ", strip=True)

    # ---- article text ----
    container = soup.find("article") or soup.find("main")

    if container is None:
        return None

    paragraphs = []
    for p in container.find_all("p"):
        txt = p.get_text(" ", strip=True)

        # ตัดบรรทัดสั้น ๆ ที่มักไม่ใช่เนื้อหาหลัก
        if len(txt) >= 30:
            paragraphs.append(txt)

    text = " ".join(paragraphs).strip()

    # ถ้าดึงมาแล้วสั้นมาก มีโอกาสว่า selector ไม่ถูก
    if len(text) < 200:
        return None

    return {
        "doc_id": normalize_url(url),
        "url": normalize_url(url),
        "title": title,
        "published_at": published_at,
        "text": text,
        "word_count": len(text.split()),
        "source_domain": urlparse(url).netloc,
    }
````

#### Step 5 — เก็บข้อมูลจริงด้วย API

หลังจากทดสอบการเรียก API และการใช้ `minTs` สำหรับโหลดข้อมูลย้อนหลังแล้ว
จึงขยายการเก็บข้อมูลข่าวอาชญากรรมจาก Amarin TV ให้มีมากกว่า 200 records
โดยกำหนด delay ระหว่าง request เพื่อไม่ส่งคำขอถี่เกินไป

````python
# ===== Step 5: Final Data Collection using API =====

API_URL = "https://web-api.amarintv.com/content/contents/list"

# cursor ที่ทดสอบแล้วว่าใช้โหลดข่าวย้อนหลังได้
before = 1790311295

all_items = []

MAX_ROUNDS = 35      # 35 รอบ × 8 ข่าว ≈ 280 ข่าว
AMOUNT = 8

for round_no in range(1, MAX_ROUNDS + 1):

    params = {
        "category": "news/crime",
        "amount": AMOUNT,
        "before": before
    }

    try:
        r = session.get(
            API_URL,
            params=params,
            timeout=20
        )

        r.raise_for_status()

        result = r.json()
        data = result.get("data", {})

        items = data.get("items", [])
        next_before = data.get("minTs")

        # เก็บข่าวจากรอบนี้
        all_items.extend(items)

        print(
            f"รอบ {round_no}:",
            f"ได้ {len(items)} ข่าว",
            f"| รวม = {len(all_items)}"
        )

        # ถ้าไม่มีข้อมูลเพิ่ม ให้หยุด
        if len(items) == 0 or next_before is None:
            print("ไม่มีข้อมูลเพิ่มเติม")
            break

        # ใช้ minTs เป็น cursor ของรอบถัดไป
        before = next_before

        # เว้นระยะเพื่อไม่ส่ง request ถี่เกินไป
        time.sleep(DELAY)

    except Exception as e:
        print(f"รอบ {round_no} ERROR:", e)
        break


# แปลงเป็น DataFrame
df_crime = pd.DataFrame(all_items)

print("\nจำนวน records ที่เก็บได้ =", len(df_crime))

# ดูตัวอย่างข้อมูล
display(
    df_crime[
        ["id", "title", "publishDate", "description", "url"
    ]].head(10)
)
````

````python
# ============================================================
# TECHNIQUE 1: COMPARISON BETWEEN GROUPS
# ============================================================

comparison = (
    news["crime_category"]
    .value_counts()
    .reset_index()
)

comparison.columns = ["crime_category", "count"]

# จำนวนข่าวทั้งหมด
total_news = comparison["count"].sum()

# คำนวณสัดส่วน %
comparison["percentage"] = (
    comparison["count"] / total_news * 100
)

# เรียงจากสัดส่วนมากไปน้อย
comparison = comparison.sort_values(
    "percentage",
    ascending=False
).reset_index(drop=True)

# เพิ่มอันดับ
comparison.insert(
    0,
    "rank",
    range(1, len(comparison) + 1)
)

print("เปรียบเทียบจำนวนและสัดส่วนข่าวแต่ละประเภท")
display(
    comparison.style.format({
        "count": "{:,}",
        "percentage": "{:.2f}%"
    })
)
````
